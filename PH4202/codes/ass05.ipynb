```python
import numpy as np
import matplotlib.pyplot as plt
# import gyrator
import scienceplots
plt.style.use('science')
plt.rcParams.update({"font.size": 25, "legend.fontsize": 15})
```

```python
# T1 = 678.86
fieldT1 = np.array([4160, 10070, 17775])
magT1 = np.array([0.649, 1.577, 2.776])

# T2 = 653.31
fieldT2 = np.array([2290, 4160, 6015, 10070, 14210, 17775, 21315])
magT2 = np.array([0.845, 1.535, 2.215, 3.665, 5.034, 6.145, 7.211])

# T3 = 629.53
fieldT3 = np.array([1355, 3230, 6015, 10070, 14210, 17775])
magT3 = np.array([6.099, 9.287, 11.793, 13.945, 15.405, 16.474])

# T4 = 614.41
fieldT4 = np.array([1820, 3230, 6015, 10070, 14210, 17775])
magT4 = np.array([19.43, 20.0, 20.77, 21.67, 22.36, 22.89])

# T5 = 601.14
fieldT5 = np.array([3230, 6015, 10070, 14210, 17775])
magT5 = np.array([25.34, 25.83, 26.4, 26.83, 27.21])
```

```python
tc = 627.2
temps = np.array([678.86, 653.31, 629.53, 614.41, 601.14])
rtemps = (tc - temps)/tc

beta = 0.368
delta = 4.22
```

```python
tempsabove = np.array([678.86, 653.31, 629.53])
rtempsabove = (tempsabove - tc)/tc

tempsbelow = np.array([614.41, 601.14])
rtempsbelow = (tc - tempsbelow)/tc
```

```python
rfieldT1 = fieldT1/(np.abs(rtemps[0])**(beta*delta))
rfieldT2 = fieldT2/(np.abs(rtemps[1])**(beta*delta))
rfieldT3 = fieldT3/(np.abs(rtemps[2])**(beta*delta))
rfieldT4 = fieldT4/(np.abs(rtemps[3])**(beta*delta))
rfieldT5 = fieldT5/(np.abs(rtemps[4])**(beta*delta))

rmagT1 = magT1/(np.abs(rtemps[0])**beta)
rmagT2 = magT2/(np.abs(rtemps[1])**beta)
rmagT3 = magT3/(np.abs(rtemps[2])**beta)
rmagT4 = magT4/(np.abs(rtemps[3])**beta)
rmagT5 = magT5/(np.abs(rtemps[4])**beta)
```

```python

fig,ax = plt.subplots(figsize=(10, 7))

ax.set_xscale('log')
ax.set_yscale('log')

ax.plot(rfieldT1, rmagT1, 'o-', label='T = 678.86 K')
ax.plot(rfieldT2, rmagT2, 'o-', label='T = 653.31 K')
ax.plot(rfieldT3, rmagT3, 'o-', label='T = 629.53 K')

ax.plot(rfieldT4, rmagT4, '-',marker = '+', markersize=10,label='T = 614.41 K')
ax.plot(rfieldT5, rmagT5, '-',marker = '+', markersize=10,label='T = 601.14 K')
ax.set_xlabel(r'$H/|t|^{\beta\delta}$')
ax.set_ylabel(r'$M/|t|^{\beta}$')
ax.legend()
fig.savefig('ass05.png', dpi=300)
```
